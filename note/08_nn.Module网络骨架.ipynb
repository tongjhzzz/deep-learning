# 神经网络的基本框架

最基本的两个函数

```python
from torch import nn
import torch

class Model(nn.Module):
    def __init__(self):
        # 一定要先初始化父类
        super().__init__()

    def forward(self, input):
        # 很简单的一个神经网络，输出只是输入加一
        output = input + 1
        return output
```

```python
# 示例使用一下
model = Model()
# 输入必须是tensor类型的
x = torch.tensor(1.0)
output = model(x)
print(output)
```
