# transforms.Resize()使用


把输入的PIL Image变成给的的尺寸，`transforms.Resize(height, width)`(在实例化对象的时候就要确定size),如果只输入了一个参数，那么就会等比缩放


# transforms.Compose()的使用


把多个变换按顺序串成一条流水线，一次调用一次执行完


```python
from PIL import Image
from torchvision import transforms

img = Image.open('/Users/daxian/deep-learning/dataset/train/bees_image/154600396_53e1252e52.jpg')

trans = transforms.Compose([
    transforms.Resize(300),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225]),
])

out = trans(img)
print(out.shape)      # torch.Size([3, 300, 421])
```

注意：

- compose列表中只需要放工具
- 可以嵌套
- 注意里面的输入输出的变化需要符合transforms的规则


```python
from PIL import Image
from torchvision import transforms

img_path = '/Users/daxian/deep-learning/dataset/train/ants_image/0013035.jpg'
img = Image.open(img_path)
print("原图 PIL.size（宽, 高）:", img.size)          # (768, 512)

# 1) 传一个整数：短边缩到 224，长边按比例缩放（保持宽高比）
r1 = transforms.Resize(224)(img)
print("Resize(224)        ->", r1.size)              # (336, 224)

# 2) 传元组：固定成 (高, 宽) = (224, 224)，宽高比会被破坏
r2 = transforms.Resize((224, 224))(img)
print("Resize((224,224))  ->", r2.size)              # (224, 224)

# 3) 元组的顺序是 (高, 宽)，所以想要 高224、宽336 要写 (224, 336)
r3 = transforms.Resize((224, 336))(img)
print("Resize((224,336))  ->", r3.size)              # (336, 224)

# 4) 放进 Compose 和后面的变换串起来
trans = transforms.Compose([
    transforms.Resize(224),
    transforms.ToTensor(),
    # Normalize 要放在 ToTensor 后面，需要时再打开
    # transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])
print("Compose 之后 tensor:", tuple(trans(img).shape))   # (3, 224, 336)
```
