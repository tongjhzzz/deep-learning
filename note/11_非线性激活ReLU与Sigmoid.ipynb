# torch.nn.ReLU

torch.nn.ReLU(inplace = False)


ReLU(x) = max(0, x) # 负数变为0，正数原样保留

原张量        : [-1.0, 2.0, -3.0]
inplace=False : 输出 [0.0, 2.0, 0.0] | 原张量还是 [-1.0, 2.0, -3.0]   ← 安全
inplace=True  : 输出 [0.0, 2.0, 0.0] | 原张量被改成 [0.0, 2.0, 0.0]   ← 原数据没了

# torch.nn.Sigmoid()

是非线性激活函数，作用是把输入压到（0， 1）区间，没有任何学习参数

两端被压的很平，中间近似线性
