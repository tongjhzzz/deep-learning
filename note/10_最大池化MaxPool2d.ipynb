# 池化

主要使用nn.MaxPool2d()

torch.nn.MaxPool2d(kernel_size, stride=None, padding=0, dilation=1, return_indices=False, ceil_mode=False)
- 需要自己填kernel_size
- stride的默认值是kernel_size
- dilation=1的情况下普通卷积，dilation = d表示核的元素之间插入(d - 1)个空格
- ceiling模式：值为小数向上取整（在池化的操作过程中就是移动步长后剩余的图片尺寸不足池化核的尺寸就保留）
- floor模式：值为小数向下取整

池化操作就是池化核对应输入取其中最大的数，目的是提取最大特征的同时减少输入量
