# 线性层

torch.nn.Linear(in_features, out_features, bias=True, device=None, dtype=None)
- in_features：输入特征数。输入张量最后一维必须等于它。
- out_features：输出特征数，也就是这一层有多少个输出神经元。
- bias=True：是否使用偏置 b。默认使用；写成 False 表示不使用偏置。
- device=None：参数存放的设备，例如 "cpu" 或 "mps"。None 表示使用默认设备。
- dtype=None：参数的数据类型，例如 torch.float32。None 表示使用默认类型。


torch.flatten(input, start_dim=0, end_dim=-1)

- input：要展平的张量
- start_dim：从第几个维度开始展平，默认是 0
- end_dim：展平到第几个维度结束，默认是最后一个维度 -1
